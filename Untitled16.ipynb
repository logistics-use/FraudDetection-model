```python
# Imports
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import time

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.svm import SVC
from sklearn.metrics import (
    classification_report, confusion_matrix,
    accuracy_score, precision_score, recall_score, f1_score,
    roc_curve, precision_recall_curve, auc
)

# Quantum computing libraries
from qiskit import QuantumCircuit, transpile
from qiskit.circuit.library import (
    PauliFeatureMap, TwoLocal
)
from qiskit.quantum_info import Statevector
from qiskit_aer import AerSimulator

# Set seed for reproducibility
np.random.seed(42)


# --- Dataset creation with fixes ---
def create_advanced_dataset(n_samples=10000, fraud_ratio=0.10, random_state=42):
    np.random.seed(random_state)
    print("Generating advanced synthetic transaction dataset...")

    n_fraud = int(np.round(n_samples * fraud_ratio))
    n_normal = n_samples - n_fraud

    # Normal transactions
    normal_amounts = np.random.lognormal(3.0, 0.5, n_normal)
    normal_hours = (np.random.vonmises(2.5, 2.0, n_normal) * 24 / (2 * np.pi)) % 24

    normal_location_probs = np.array([0.1]*10 + [0.05]*5 + [0.01]*5, dtype=float)
    normal_location_probs = normal_location_probs / normal_location_probs.sum()
    normal_locations = np.random.choice(np.arange(1, 21), size=n_normal, p=normal_location_probs)

    normal_merchant_probs = np.array(
        [0.2, 0.15, 0.1, 0.1, 0.08, 0.07, 0.06, 0.05, 0.04,
         0.03, 0.03, 0.02, 0.02, 0.02, 0.01], dtype=float
    )
    normal_merchant_probs = normal_merchant_probs / normal_merchant_probs.sum()
    normal_merchants = np.random.choice(np.arange(1, 16), size=n_normal, p=normal_merchant_probs)

    normal_behavior = np.random.beta(2, 5, n_normal)

    # Fraud transactions
    fraud_types = np.random.choice([1, 2, 3], n_fraud, p=[0.5, 0.3, 0.2])

    fraud_amounts = []
    fraud_hours = []
    fraud_locations = []
    fraud_merchants = []
    fraud_behavior = []

    for t in fraud_types:
        if t == 1:  # Large purchase fraud
            fraud_amounts.append(np.random.lognormal(5.5, 0.7))
            fraud_hours.append(np.random.uniform(1, 5))
            fraud_locations.append(np.random.choice(np.arange(18, 26)))
            fraud_merchants.append(np.random.choice([12, 13, 14, 15]))
            fraud_behavior.append(np.random.beta(1, 10))
        elif t == 2:  # Micro transaction fraud
            fraud_amounts.append(np.random.uniform(0.1, 2.0))
            fraud_hours.append(np.random.uniform(0, 24))
            fraud_locations.append(np.random.choice(np.arange(1, 26)))
            fraud_merchants.append(np.random.choice([8, 9, 10, 11]))
            fraud_behavior.append(np.random.beta(1, 8))
        else:  # Behavioral fraud
            fraud_amounts.append(np.random.lognormal(3.5, 0.6))
            fraud_hours.append((np.random.vonmises(0.5, 1.5) * 24 / (2 * np.pi)) % 24)
            fraud_locations.append(np.random.choice(np.arange(15, 26)))
            fraud_merchants.append(np.random.choice(np.arange(1, 16)))
            fraud_behavior.append(np.random.beta(1, 15))

    amounts = np.concatenate([normal_amounts, np.array(fraud_amounts)])
    hours = np.concatenate([normal_hours, np.array(fraud_hours)])
    locations = np.concatenate([normal_locations, np.array(fraud_locations)])
    merchants = np.concatenate([normal_merchants, np.array(fraud_merchants)])
    behavior = np.concatenate([normal_behavior, np.array(fraud_behavior)])

    amount_zscore = (amounts - amounts.mean()) / (amounts.std() + 1e-9)
    time_from_avg = np.abs(hours - hours.mean())
    location_rarity = np.array([
        0.1 if loc <= 10 else 0.05 if loc <= 15 else 0.01 for loc in locations
    ])
    merchant_risk = np.array([
        0.01 if m <= 5 else 0.05 if m <= 10 else 0.1 for m in merchants
    ])

    X = np.column_stack([
        amounts, hours, locations, merchants, behavior,
        amount_zscore, time_from_avg, location_rarity, merchant_risk
    ])

    y = np.concatenate([np.zeros(n_normal, dtype=int), np.ones(n_fraud, dtype=int)])
    fraud_categories = np.concatenate([np.zeros(n_normal, dtype=int), fraud_types])

    idx = np.arange(X.shape[0])
    np.random.shuffle(idx)
    X = X[idx]
    y = y[idx]
    fraud_categories = fraud_categories[idx]

    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    print(f"Dataset created with {X_scaled.shape[0]} samples, {X_scaled.shape[1]} features")
    print(f"Fraud distribution: {y.sum()} fraud cases ({y.sum()/len(y)*100:.4f}%)")

    return X_scaled, y, fraud_categories, scaler


# --- Quantum angle encoder with guard for constant features ---
class QuantumAngleEncoder:
    def __init__(self, n_qubits, encoding_type='ry'):
        self.n_qubits = n_qubits
        self.encoding_type = encoding_type
        self.pca = None
        self.normalization_params = None

    def fit_transform(self, X):
        print("Performing advanced quantum angle encoding...")
        self.feature_ranges = [(X[:, i].min(), X[:, i].max()) for i in range(X.shape[1])]

        if X.shape[1] > self.n_qubits:
            self.pca = PCA(n_components=self.n_qubits)
            X_reduced = self.pca.fit_transform(X)
            print(f"Reduced features from {X.shape[1]} to {self.n_qubits} using PCA")
        else:
            X_reduced = X

        X_min = X_reduced.min(axis=0)
        X_max = X_reduced.max(axis=0)
        denom = X_max - X_min
        denom[denom == 0] = 1e-9  # avoid division by zero

        X_normalized = (X_reduced - X_min) / denom * np.pi
        self.normalization_params = (X_min, X_max)

        return X_normalized

    def transform(self, X):
        if self.pca is not None:
            X_reduced = self.pca.transform(X)
        else:
            X_reduced = X

        X_min, X_max = self.normalization_params
        denom = X_max - X_min
        denom[denom == 0] = 1e-9
        X_normalized = (X_reduced - X_min) / denom * np.pi

        return X_normalized

    def create_encoding_circuit(self, angles, draw_circuit=True):
        n_qubits = min(len(angles), self.n_qubits)
        qc = QuantumCircuit(n_qubits)

        if self.encoding_type == 'ry':
            for i in range(n_qubits):
                qc.ry(angles[i], i)
        elif self.encoding_type == 'rzry':
            for i in range(n_qubits):
                qc.rz(angles[i], i)
                qc.ry(angles[i] / 2, i)

        for i in range(n_qubits - 1):
            qc.cx(i, i + 1)

        if draw_circuit:
            try:
                fig, ax = plt.subplots(figsize=(10, 5))
                qc.draw('mpl', ax=ax)
                ax.set_title(f'Quantum Angle Encoding Circuit ({self.encoding_type} encoding)')
                plt.tight_layout()
                plt.show()
            except Exception as e:
                print(f"Circuit visualization not available: {e}")
                print(qc)

        return qc

    def visualize_bloch_sphere(self, angles, title="Bloch Sphere Representation"):
        qc = self.create_encoding_circuit(angles, draw_circuit=False)
        try:
            simulator = AerSimulator()
            transpiled_qc = transpile(qc, simulator)
            result = simulator.run(transpiled_qc).result()
            statevector = result.get_statevector()
        except Exception as e:
            # fallback to classical statevector
            statevector = Statevector.from_instruction(qc)
            print(f"Using Statevector fallback: {e}")

        try:
            from qiskit.visualization import plot_bloch_multivector
            fig = plot_bloch_multivector(statevector, title=title)
            plt.show()
        except Exception as e:
            print(f"Bloch sphere visualization not available: {e}")

        return statevector


# --- Simple quantum‐kernel fallback and hybrid models remain largely the same ---
class SimpleQuantumKernel:
    def __init__(self, feature_map):
        self.feature_map = feature_map

    def evaluate(self, X, Y=None):
        if Y is None:
            Y = X
        n_x = X.shape[0]
        n_y = Y.shape[0]
        K = np.zeros((n_x, n_y))
        for i in range(n_x):
            for j in range(n_y):
                # rbf‐like kernel fallback
                K[i, j] = np.exp(-0.5 * np.linalg.norm(X[i] - Y[j])**2)
        return K


class AdvancedVQC:
    def __init__(self, n_qubits, feature_map_reps=2, ansatz_reps=3):
        self.n_qubits = n_qubits
        self.feature_map_reps = feature_map_reps
        self.ansatz_reps = ansatz_reps
        self.kernel = None
        self.model = None

        self.feature_map = PauliFeatureMap(
            feature_dimension=n_qubits,
            reps=feature_map_reps,
            paulis=['Z', 'ZZ', 'X', 'XX']
        )
        self.ansatz = TwoLocal(
            num_qubits=n_qubits,
            rotation_blocks=['ry', 'rz'],
            entanglement_blocks='cx',
            entanglement='circular',
            reps=ansatz_reps,
            skip_final_rotation_layer=False
        )
        self.circuit = QuantumCircuit(n_qubits)
        self.circuit.compose(self.feature_map, inplace=True)
        self.circuit.compose(self.ansatz, inplace=True)

    def visualize_circuit(self, feature_map_only=False, ansatz_only=False):
        if feature_map_only:
            circ = self.feature_map
            title = "Feature Map Circuit"
        elif ansatz_only:
            circ = self.ansatz
            title = "Ansatz Circuit"
        else:
            circ = self.circuit
            title = "Full VQC Circuit"
        try:
            fig, ax = plt.subplots(figsize=(14, 8))
            circ.decompose().draw('mpl', ax=ax, style='clifford')
            ax.set_title(title, fontsize=16)
            plt.tight_layout()
            plt.show()
        except Exception as e:
            print(f"Circuit visualization not available: {e}")
            print(circ)
        print(f"{title} info: qubits={circ.num_qubits}, parameters={circ.num_parameters}, depth={circ.depth()}")

    def train(self, X_train, y_train):
        print("Training Advanced VQC Model...")

        self.kernel = SimpleQuantumKernel(self.feature_map)
        kernel_func = self.kernel.evaluate

        self.model = SVC(
            kernel=kernel_func,
            C=1.0,
            class_weight='balanced',
            probability=True
        )

        start_time = time.time()
        self.model.fit(X_train, y_train)
        training_time = time.time() - start_time
        print(f"VQC training completed in {training_time:.2f} seconds")
        return self

    def predict(self, X):
        return self.model.predict(X)

    def predict_proba(self, X):
        return self.model.predict_proba(X)

    def decision_function(self, X):
        return self.model.decision_function(X)


class AdvancedQSVM:
    def __init__(self, n_qubits, feature_map_reps=3):
        self.n_qubits = n_qubits
        self.feature_map_reps = feature_map_reps
        self.kernel = None
        self.model = None

        self.feature_map = TwoLocal(
            num_qubits=n_qubits,
            reps=feature_map_reps,
            entanglement='full',
            rotation_blocks=['ry', 'rz']
        )

    def visualize_circuit(self):
        try:
            fig, ax = plt.subplots(figsize=(14, 6))
            self.feature_map.decompose().draw('mpl', ax=ax, style='clifford')
            ax.set_title("QSVM Feature Map Circuit", fontsize=16)
            plt.tight_layout()
            plt.show()
        except Exception as e:
            print(f"Circuit visualization not available: {e}")
            print(self.feature_map)
        print(f"QSVM Feature Map info: qubits={self.feature_map.num_qubits}, parameters={self.feature_map.num_parameters}, depth={self.feature_map.depth()}")

    def train(self, X_train, y_train):
        print("Training Advanced QSVM Model for Anomaly Detection...")
        self.kernel = SimpleQuantumKernel(self.feature_map)
        kernel_func = self.kernel.evaluate

        self.model = SVC(
            kernel=kernel_func,
            C=1.0,
            class_weight='balanced',
            probability=True
        )

        start_time = time.time()
        self.model.fit(X_train, y_train)
        training_time = time.time() - start_time
        print(f"QSVM training completed in {training_time:.2f} seconds")
        return self

    def predict(self, X):
        return self.model.predict(X)

    def predict_proba(self, X):
        return self.model.predict_proba(X)

    def decision_function(self, X):
        return self.model.decision_function(X)


class QuantumEnsemble:
    def __init__(self, models, weights=None):
        self.models = models
        self.weights = weights if weights else [1.0/len(models)] * len(models)

    def predict_proba(self, X):
        probas = []
        for i, mdl in enumerate(self.models):
            pr = mdl.predict_proba(X)
            probas.append(pr * self.weights[i])
        total = np.sum(probas, axis=0)
        return total

    def predict(self, X, threshold=0.5):
        proba = self.predict_proba(X)
        return (proba[:, 1] > threshold).astype(int)

    def decision_function(self, X):
        decisions = []
        for i, mdl in enumerate(self.models):
            if hasattr(mdl, 'decision_function'):
                d = mdl.decision_function(X)
                decisions.append(d * self.weights[i])
        return np.sum(decisions, axis=0)


# --- Main execution with baseline and minor tweaks ---
def main():
    print("=== Advanced Quantum Hybrid Fraud Detection System ===")

    X, y, fraud_categories, scaler = create_advanced_dataset(
        n_samples=5000,  # reduce size for quicker runs
        fraud_ratio=0.05
    )

    X_train, X_test, y_train, y_test, cat_train, cat_test = train_test_split(
        X, y, fraud_categories,
        test_size=0.3,
        random_state=42,
        stratify=y
    )

    n_qubits = 4
    encoder = QuantumAngleEncoder(n_qubits, encoding_type='ry')
    X_train_encoded = encoder.fit_transform(X_train)
    X_test_encoded = encoder.transform(X_test)

    print(f"\nOriginal X shape: {X.shape}")
    print(f"Encoded X_train shape: {X_train_encoded.shape}")

    # Baseline classical SVM (RBF)
    print("\n--- Training classical baseline SVM ---")
    baseline_svm = SVC(kernel='rbf', probability=True, class_weight='balanced')
    start = time.time()
    baseline_svm.fit(X_train, y_train)
    print(f"Baseline SVM training time: {time.time() - start:.2f}s")

    # Train quantum/hybrid models
    print("\n--- Training VQC model ---")
    vqc_model = AdvancedVQC(n_qubits, feature_map_reps=2, ansatz_reps=2)
    vqc_model.visualize_circuit(feature_map_only=True)
    vqc_model.train(X_train_encoded, y_train)

    print("\n--- Training QSVM model ---")
    qsvm_model = AdvancedQSVM(n_qubits, feature_map_reps=2)
    qsvm_model.visualize_circuit()
    qsvm_model.train(X_train_encoded, y_train)

    ensemble = QuantumEnsemble(models=[vqc_model, qsvm_model], weights=[0.6, 0.4])

    # Predictions
    baseline_pred = baseline_svm.predict(X_test)
    baseline_proba = baseline_svm.predict_proba(X_test)

    vqc_pred = vqc_model.predict(X_test_encoded)
    vqc_proba = vqc_model.predict_proba(X_test_encoded)

    qsvm_pred = qsvm_model.predict(X_test_encoded)
    qsvm_proba = qsvm_model.predict_proba(X_test_encoded)

    ensemble_pred = ensemble.predict(X_test_encoded)
    ensemble_proba = ensemble.predict_proba(X_test_encoded)

    # Evaluation loop
    print("\n=== Model Performance Evaluation ===")
    models = {
        'Baseline SVM': (baseline_pred, baseline_proba),
        'VQC': (vqc_pred, vqc_proba),
        'QSVM': (qsvm_pred, qsvm_proba),
        'Ensemble': (ensemble_pred, ensemble_proba)
    }

    for name, (pred, proba) in models.items():
        acc = accuracy_score(y_test, pred)
        prec = precision_score(y_test, pred, zero_division=0)
        rec = recall_score(y_test, pred, zero_division=0)
        f1 = f1_score(y_test, pred, zero_division=0)
        print(f"\n{name}: Accuracy={acc:.4f}, Precision={prec:.4f}, Recall={rec:.4f}, F1={f1:.4f}")
        cm = confusion_matrix(y_test, pred)
        plt.figure(figsize=(5,4))
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                    xticklabels=['Normal','Fraud'], yticklabels=['Normal','Fraud'])
        plt.title(f'Confusion Matrix — {name}')
        plt.ylabel('True label')
        plt.xlabel('Predicted label')
        plt.show()

    # ROC & Precision‐Recall plots
    plt.figure(figsize=(14,6))
    plt.subplot(1,2,1)
    for name, (_, proba) in models.items():
        fpr, tpr, _ = roc_curve(y_test, proba[:,1])
        roc_auc = auc(fpr, tpr)
        plt.plot(fpr, tpr, label=f'{name} (AUC={roc_auc:.3f})')
    plt.plot([0,1], [0,1], 'k--')
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('ROC Curves')
    plt.legend()

    plt.subplot(1,2,2)
    for name, (_, proba) in models.items():
        precision_vals, recall_vals, _ = precision_recall_curve(y_test, proba[:,1])
        pr_auc = auc(recall_vals, precision_vals)
        plt.plot(recall_vals, precision_vals, label=f'{name} (AUC={pr_auc:.3f})')
    plt.xlabel('Recall')
    plt.ylabel('Precision')
    plt.title('Precision–Recall Curves')
    plt.legend()
    plt.tight_layout()
    plt.show()

    # Case analysis (few samples)
    print("\n=== Case Analysis ===")
    fraud_idx = np.where(y_test == 1)[0][:3]
    normal_idx = np.where(y_test == 0)[0][:3]

    print("\nFraud case analysis:")
    for i, idx in enumerate(fraud_idx, start=1):
        print(f"\nFraud Case {i}: True label = Fraud (Category = {cat_test[idx]})")
        print(f"Baseline SVM proba: {baseline_proba[idx,1]:.3f}")
        print(f"VQC proba: {vqc_proba[idx,1]:.3f}")
        print(f"QSVM proba: {qsvm_proba[idx,1]:.3f}")
        print(f"Ensemble proba: {ensemble_proba[idx,1]:.3f}")
        print(f"Ensemble prediction: {'Fraud' if ensemble_pred[idx]==1 else 'Normal'}")

    print("\nNormal case analysis:")
    for i, idx in enumerate(normal_idx, start=1):
        print(f"\nNormal Case {i}: True label = Normal")
        print(f"Baseline SVM proba: {baseline_proba[idx,1]:.3f}")
        print(f"VQC proba: {vqc_proba[idx,1]:.3f}")
        print(f"QSVM proba: {qsvm_proba[idx,1]:.3f}")
        print(f"Ensemble proba: {ensemble_proba[idx,1]:.3f}")
        print(f"Ensemble prediction: {'Fraud' if ensemble_pred[idx]==1 else 'Normal'}")

    return {
        'baseline_svm': baseline_svm,
        'vqc_model': vqc_model,
        'qsvm_model': qsvm_model,
        'ensemble': ensemble,
        'encoder': encoder,
        'results': {name: accuracy_score(y_test, models[name][0]) for name in models}
    }


if __name__ == "__main__":
    results = main()
```
